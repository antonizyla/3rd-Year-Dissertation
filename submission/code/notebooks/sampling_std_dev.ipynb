```python
from Influence_Modelling import Sampling, XsvNetwork, Metrics

import pandas as pd
import numpy as np

enron = XsvNetwork.XsvNetwork("enron")
deviations = np.zeros(len([x for x in range(2, 20)]))
for i in range(10):
  rows = []
  std_edges = {}
  for i in range(2, 20):
    row = Metrics.generate_tabular_row(Sampling.mhda_sample(enron.get_nx(), 1000))
    rows.append(row)
    df = pd.concat(rows)
    std_edges[i] = df["Edges"].std()
  std_edges= list(std_edges.items())
  number_samples = [x for (x, _) in std_edges]
  edges_deviation = np.array([x for (_, x) in std_edges])
  deviations += edges_deviation
deviations /= len([x for x in range(2, 20)])
```

```python
import matplotlib.pyplot as plt
import scienceplots
plt.style.use(["science", "no-latex"])

fig, ax = plt.subplots()
fig.set_size_inches(5,4)
fig.set_dpi(200)
ax.set_title("Standard Deviation of Edges (10 trials)")
ax.set_xlabel("Number of samples")
ax.set_ylabel("Mean standard deviation")
ax.plot([x for x in range(2, 20)], deviations)
```

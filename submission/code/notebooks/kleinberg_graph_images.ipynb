```python
import networkx as nx
from networkx.drawing.nx_pydot import graphviz_layout
import matplotlib.pyplot as plt
import scienceplots
plt.style.use(["science", "no-latex"])

G = nx.navigable_small_world_graph(3)
G = G.to_undirected()
pos = {node: node for node in G.nodes()}

fig, axs = plt.subplots()
fig.set_size_inches(5,4)
nx.draw(G, pos, with_labels=False)
plt.title("Example 3 by 3 Navigable Small World Graph")
plt.show()
fig.savefig("../latex_docs/progress/example_3by3_kleinberg_graph.png", format="png")
```

```python
from Influence_Modelling import XsvNetwork
p2p = XsvNetwork.XsvNetwork("airports")
pos = graphviz_layout(p2p.get_nx(), "sfdp")
```

```python
fig, ax = plt.subplots()
fig.set_size_inches(9, 9)
fig.set_dpi(1000)
ax.set_title("Airports Graph edges visualised (Graphviz with sfdp)")
nx.draw_networkx_edges(p2p.get_nx(), pos, width=0.1, alpha=0.7)
fig.savefig("../latex_docs/progress/airports_graph_edges_image.png", format="png")
```

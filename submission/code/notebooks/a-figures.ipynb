```python
import pickle 
import matplotlib.pyplot as plt
import scienceplots
plt.style.use(["science", "no-latex"])

def read(filepath):
    with open(filepath, "rb") as file_handle:
        return pickle.load(file_handle)

all_tables = read("/dcs/23/u5527350/3rdyearproject/results/ICC-25/All-ICC-Compiled-Tables.pickle")

small = all_tables[0:9]
large = all_tables[9:]
```

```python
markers =  ['o', 's', '^', 'D', 'v', '>', '<', 'p', '*', 'x']

fig, axs = plt.subplots(1, 2)
fig.suptitle("Heuristics Compared on ICC Model (250 trials)")
fig.set_size_inches(12, 6)
legends = []
for i, table in enumerate(small):
    table = table.iloc[::-1]
    axs[0].plot(table['n'].to_numpy(), table['end_coverage'], linewidth=0.5, marker=markers[i % len(markers)])
    legends.append(table['heuristic'].unique()[0])
for i, table in enumerate(large):
    table = table.iloc[::-1]
    legends.append(table['heuristic'].unique()[0])
    axs[1].plot(table['n'].to_numpy(), table['end_coverage'], linewidth=0.5, marker=markers[i % len(markers)])
    
axs[1].set_xlabel("Number of Seed Nodes")
axs[1].set_ylabel("Infected Proprotion at End")
axs[1].set_title("MHDA Enron 4096")
axs[0].set_xlabel("Number of Seed Nodes")
axs[0].set_ylabel("Infected Proprotion at End")
axs[0].set_title("MHDA Enron 1024")
axs[0].legend(legends)
axs[1].legend(legends)

plt.savefig("../all_heuristics_icc_enron_mhda_2_sizes", format="png")
```

```python
import numpy as np
markers = ['o', 's', '^', 'D', 'v', '>', '<', 'p', '*', 'x']

# work on generating figures that are relative to one another
fig, axs = plt.subplots(1, 2)
fig.suptitle("Heuristics Compared on ICC Model against greedy as baseline (250 trials)")
fig.set_size_inches(12, 6)

baseline_small = small[8]
print(f"Baseline Heuristic for 1024: {baseline_small.iloc[0]['heuristic']}")
baseline_small = baseline_small.iloc[::-1]
baseline_large = large[8]
print(f"Baseline Heuristic for 4096: {baseline_large.iloc[0]['heuristic']}")
baseline_large = baseline_large.iloc[::-1]

legends = []
for i, table in enumerate(small):
    table = table.iloc[::-1]
    axs[0].plot(table['n'].to_numpy(), table['end_coverage'] - baseline_small['end_coverage'], linewidth=0.5, marker=markers[i % len(markers)])
    legends.append(table['heuristic'].unique()[0])
for i, table in enumerate(large):
    table = table.iloc[::-1]
    legends.append(table['heuristic'].unique()[0])
    axs[1].plot(table['n'].to_numpy(), table['end_coverage'] - baseline_large['end_coverage'], linewidth=0.5, marker=markers[i % len(markers)])
    
axs[1].set_xlabel("Number of Seed Nodes")
axs[1].set_ylabel("Difference between Random Infected Proportion")
axs[1].set_title("MHDA-Sampled Enron 4096")
axs[0].set_xlabel("Number of Seed Nodes")
axs[0].set_ylabel("Difference between Random Infected Proportion")
axs[0].set_title("MHDA-Sampled Enron 1024")
axs[0].legend(legends)
axs[1].legend(legends)

plt.savefig("../all_heuristics_random_as_baseline_mhda_enron_2_sizes.png")
```

```python
colours = ["blue", "green", "orange", "red", "purple", "black", "gray", "blue", "green"]
small_tuples = []
for i, data in enumerate(small):
    small_tuples.append(
        (
            np.log2(data['time_taken_ns'].mean()/1000), 
            data['heuristic'][0].unique()[0], 
            colours[i % len(small)]
        )
    )
small_tuples.sort()

large_tuples = []
for i, data in enumerate(large):
    large_tuples.append(
        (
            np.log2(data['time_taken_ns'].mean()/1000), 
            data['heuristic'][0].unique()[0],
            colours[i % len(large)]
        )
    )
large_tuples.sort()

fig, axs = plt.subplots(1,2)
fig.set_size_inches(12,6)

axs[0].barh([name for _, name, _ in small_tuples], [val for val, _, _ in small_tuples], color=[col for _, _, col in small_tuples])
axs[0].set_xlabel("log base 2 of mean time taken (ms)")
axs[0].set_title("Enron 1024 MHDA")
axs[1].barh([name for _, name, _ in large_tuples], [val for val, _, _ in large_tuples], color=[col for _, _, col in large_tuples])
axs[1].set_xlabel("log base 2 of mean time taken (ms)")
axs[1].set_title("Enron 4096 MHDA")

plt.savefig("../heuristic_runtime_compared.png", format="png")
```

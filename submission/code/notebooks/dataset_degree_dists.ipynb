```python
import networkx as nx
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from Influence_Modelling import XsvNetwork
import scienceplots
plt.style.use(["science", "no-latex"])

def plot_degree_histogram(G, axis, title):
  axis.hist([G.degree(g) for g in G.nodes()], log=True)
  axis.set_xlabel("Degree")
  axis.set_ylabel("Number of Nodes")
  axis.set_title(f"Degree Distribution {title}")
  axis.legend()
```

```python
fig, ax = plt.subplots()
enron = XsvNetwork.XsvNetwork("enron")
plot_degree_histogram(enron.get_nx(), ax, "Enron")
fig.set_dpi(200)
fig.set_size_inches(4, 4)
plt.savefig("Enron_whole_deg_hist.png", format="png")
```

```python
fig, ax = plt.subplots()
twitter = XsvNetwork.XsvNetwork("twitter")
plot_degree_histogram(twitter.get_nx(), ax, "Twitter")
fig.set_dpi(200)
fig.set_size_inches(4, 4)
plt.savefig("Twitter_whole_deg_hist.png", format="png")
```

```python
fig, ax = plt.subplots()
p2p = XsvNetwork.XsvNetwork("p2p")
plot_degree_histogram(p2p.get_nx(), ax, "P2P")
fig.set_dpi(200)
fig.set_size_inches(4, 4)
plt.savefig("p2p_whole_deg_hist.png", format="png")
```

```python
fig, ax = plt.subplots()
airports = XsvNetwork.XsvNetwork("airports")
plot_degree_histogram(airports.get_nx(), ax, "airports")
fig.set_dpi(200)
fig.set_size_inches(4, 4)
plt.savefig("Airports_whole_deg_hist.png", format="png")
```

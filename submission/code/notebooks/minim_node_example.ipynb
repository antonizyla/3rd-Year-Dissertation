```python
import networkx as nx 
import matplotlib.pyplot as plt 
import scienceplots
plt.style.use(["science","no-latex"])
import numpy as np
import pandas as pd 
from Influence_Modelling import XsvNetwork, Network
from Influence_Modelling import Heuristics, Sampling

en = XsvNetwork.XsvNetwork("enron")
net = Network.Network("")
net.set_nx(Sampling.mhda_sample(en.get_nx(), 1024))
```

```python
inf = np.linspace(0.1, 0.9, 9)
deg = []
local_clustering = []
discount = []
random = []
for j in inf:
  deg.append(net.run_ICC_find_min_nodes(Heuristics.deg_h, float(j)))
  discount.append(net.run_ICC_find_min_nodes(Heuristics.deg_discount_h, float(j)))
  # I have 0 idea as to why this is not working.
  #local_clustering.append(net.run_ICC_find_min_nodes(Heuristics.local_clustering_coefficient_h, float(j)))
  #random.append(net.run_ICC_find_min_nodes(Heuristics.random_h, float(j)))
```

```python
fig, ax = plt.subplots()
fig.set_size_inches(6, 5)
ax.set_title("Minimum Number of Nodes ")

ax.plot(inf, deg, marker="o", label="Degree")
ax.plot(inf, local_clustering, label="Local Clustering Coef.", marker="^")
ax.plot(inf, discount, label="Degree Discount", marker="v")
ax.plot
ax.set_xlabel("Desired Proportion")
ax.set_ylabel("Required Number of Nodes")
ax.legend()

plt.savefig("/home/antoni/Dev/3rdyearproject/latex_docs/progress/minimum_nodes_compared.png", format="png")
```

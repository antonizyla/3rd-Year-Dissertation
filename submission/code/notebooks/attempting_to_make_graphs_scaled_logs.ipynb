```python
ICC = "/dcs/23/u5527350/3rdyearproject/results/ICC-run2/All-ICC-Compiled-Tables.pickle"
DCC = "/dcs/23/u5527350/3rdyearproject/results/DCC-run2/All-DCC-Compiled-Tables.pickle"
LTM = "/dcs/23/u5527350/3rdyearproject/results/LTM-run2/All-LTM-Compiled-Tables.pickle"
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import scienceplots
plt.style.use(['science', 'no-latex'])
import pickle
import pandas as pd
results = []
with open(ICC, "rb") as f:
    results.append(pickle.load(f))
with open(DCC, "rb") as f:
    results.append(pickle.load(f))
with open(LTM, "rb") as f:
    results.append(pickle.load(f))
```

```python
ICC_Sampled = [i for i in results[0] if i['dataset'].iloc[0] == 'Enron-MHDA-1024']
ICC_Unsampled = [i for i in results[0] if i['dataset'].iloc[0] != 'Enron-MHDA-1024']
DCC_Sampled = [i for i in results[1] if i['dataset'].iloc[0] == 'Enron-MHDA-1024']
DCC_Unsampled = [i for i in results[1] if i['dataset'].iloc[0] != 'Enron-MHDA-1024']
LTM_Sampled = [i for i in results[2] if i['dataset'].iloc[0] == 'Enron-MHDA-1024']
LTM_Unsampled = [i for i in results[2] if i['dataset'].iloc[0] != 'Enron-MHDA-1024']
plt.rcParams['figure.figsize'] = [10, 14]
fig, axs = plt.subplots(3,2)
fig.suptitle("Heuristics Compared (50 Runs - 29mins on 64 cores)")
legends = []
for table in ICC_Sampled:
    table = table.iloc[::-1]
    axs[0, 0].plot(table['n'].to_numpy(), np.log2((table['end_coverage']-table['start_coverage'])/table['start_coverage']) )
    legends.append(table['heuristic'].unique()[0])
    axs[0, 0].set_title("ICC MHDA 1024")
for table in ICC_Unsampled:
    table = table.iloc[::-1]
    axs[0, 1].plot(table['n'].to_numpy(), np.log2((table['end_coverage']-table['start_coverage'])/table['start_coverage']))
    legends.append(table['heuristic'].unique()[0])
    axs[0, 1].set_title("ICC MHDA 4096")
for table in DCC_Sampled:
    table = table.iloc[::-1]
    axs[1, 0].plot(table['n'].to_numpy(), np.log2((table['end_coverage']-table['start_coverage'])/table['start_coverage']))
    legends.append(table['heuristic'].unique()[0])
    axs[1, 0].set_title("DCC MHDA 1024")
for table in DCC_Unsampled:
    table = table.iloc[::-1]
    axs[1, 1].plot(table['n'].to_numpy(),np.log2((table['end_coverage']-table['start_coverage'])/table['start_coverage']))
    legends.append(table['heuristic'].unique()[0])
    axs[1, 1].set_title("DCC MHDA 4096")
for table in LTM_Sampled:
    table = table.iloc[::-1]
    axs[2, 0].plot(table['n'].to_numpy(), np.log2((table['end_coverage']-table['start_coverage'])/table['start_coverage']))
    legends.append(table['heuristic'].unique()[0])
    axs[2, 0].set_title("LTM MHDA 1024")
for table in LTM_Unsampled:
    table = table.iloc[::-1]
    axs[2, 1].plot(table['n'].to_numpy(), np.log2((table['end_coverage']-table['start_coverage'])/table['start_coverage']))
    legends.append(table['heuristic'].unique()[0])
    axs[2, 1].set_title("LTM MHDA 4096")
for ax in axs.flat:
    ax.set_xlabel("Number of Seed Nodes")
    ax.set_ylabel("log2((end-start)/start)")
    ax.legend(legends)
plt.show()
fig.savefig("example-plots-2")
```

```python
def degree_h(G: nx.Graph, seed_size: int = 2):
    # choose the highest degree vertices for a seed
    ordered_nodes = []
    for node in G.nodes():
        ordered_nodes.append((G.degree(node), node))
    ordered_nodes.sort(reverse=True)
    return [x[1] for x in ordered_nodes[:seed_size]]


def degree_discount_h(G: nx.Graph, seed_size: int = 2):
    # choose vertices based on maximal degree
    # if a vertex v is chosen then all neighbours of v degree's will decrease by 1 and the next highest will be chosen
    vertices = []
    while seed_size != 0:
        degrees = G.degree()
        maximal_deg = max([v for (u,v) in degrees])
        vertex = [u for (u,v) in degrees if v == maximal_deg][0]
        vertices.append(vertex)
        G.remove_node(vertex)
        seed_size -= 1
    return vertices
```

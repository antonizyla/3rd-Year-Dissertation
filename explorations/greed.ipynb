```python
import networkx as nx
from Influence_Modelling import Network, XsvNetwork, Sampling

def greedy2_h(G: nx.DiGraph, n: int = 2) -> list:
    if n >= len(G.nodes()):
        return list(G.nodes())

    net = Network.Network("")
    net.set_nx(G)
    seeds = []
    for _ in range(n):
        print("Got a node")
        best_node = None
        best_spread = -1
        for node in G.nodes():
            if node in seeds:
                continue
            candidate = seeds + [node]
            _, final, _ = net.run_ICC(candidate, 0.1, 10000)
            spread = sum(
                1 for _, data in final.nodes(data=True)
                if data["status"] != "inactive"
            )
            spread /= final.number_of_nodes()
            if spread > best_spread:
                best_spread = spread
                best_node = node
        seeds.append(best_node)
    return seeds

p2p = XsvNetwork.XsvNetwork("p2p")
greedy2_h(Sampling.mhda_sample(p2p.get_nx(), 1000), 100)
```

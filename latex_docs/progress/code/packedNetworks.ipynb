This is the code used to generate the table that shows key metrics about the loaded networks

```python
import pandas as pd
from Influence_Modelling import XsvNetwork, NetworkMetrics

nets = [
    XsvNetwork.XsvNetwork("", "", "enron"),
    XsvNetwork.XsvNetwork("", "", "p2p"),
    XsvNetwork.XsvNetwork("", "", "airports"),
]

table = []
for net in nets:
    n = net.get_nx()
    table.append(
        pd.DataFrame(
            [
                [
                    "",
                    NetworkMetrics.s_1_number_nodes(n),
                    NetworkMetrics.s_2_number_edges(n),
                    NetworkMetrics.s_3_mean_degree(n),
                    NetworkMetrics.s_4_clustering_coefficient(n),
                    NetworkMetrics.s_5_diameter(n),
                    NetworkMetrics.s_6_centralization(n),
                ]
            ]
        )
    )
```

```python
t = pd.concat(table)
t.columns = [
    "Network",
    "Node",
    "Edges",
    "Mean Degree",
    "Clustering Coefficient",
    "Diameter",
    "Centralization",
]
t
```
